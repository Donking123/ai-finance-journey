## Part 4: Filter Rows

New concepts:
- `df[df["sector"] == "Banking"]` — keep rows where sector equals "Banking"
- `df[df["dividend_yield"] > 5.0]` — keep rows where yield is above 5
- `df["sector"].str.contains("REIT")` — keep rows where sector contains "REIT"

How filtering works:
1. `df["sector"] == "Banking"` produces True/False for every row
2. `df[True/False list]` keeps only the True rows

```python
# TODO: Filter — only banking stocks
#   banks = df[df["sector"] == "Banking"]
#   banks[["name", "price", "dividend_yield"]]
df[df["sector"] == "Banking"]
```

```python
# TODO: Filter — stocks with dividend yield above 5%
#   high_yield = df[df["dividend_yield"] > 5.0]
#   high_yield[["name", "sector", "dividend_yield"]]
df[df["dividend_yield"] > 5.0]
```

```python
# TODO: Filter — REITs only (sector contains "REIT")
#   reits = df[df["sector"].str.contains("REIT")]
#   reits[["name", "sector", "dividend_yield"]]
df[df["sector"].str.contains("REIT")]
```

## Part 5: Sort Values

New concepts:
- `df.sort_values("column")` — sort ascending (smallest first)
- `df.sort_values("column", ascending=False)` — sort descending (largest first)
- Chaining: `df.sort_values(...).head()` — sort, then take top 5

```python
df.head(3)
```

```python
# TODO: Top 5 by market cap (largest first)
#   df.sort_values("market_cap_m", ascending=False)[["name", "sector", "market_cap_m"]].head()
df.sort_values("market_cap_m", ascending=False)[["name", "sector", "market_cap_m"]].head()
```

```python
# TODO: Top 5 by dividend yield (highest first)
#   df.sort_values("dividend_yield", ascending=False)[["name", "sector", "dividend_yield"]].head()
df.sort_values("dividend_yield", ascending=False)[["name", "sector", "dividend_yield"]].head()
```

```python
# TODO: Top 5 cheapest by PE ratio (lowest first, exclude 0)
#   has_pe = df[df["pe_ratio"] > 0]
#   has_pe.sort_values("pe_ratio")[["name", "sector", "pe_ratio"]].head()
has_pe = df[df["pe_ratio"] > 0]
has_pe.sort_values("pe_ratio")[["name", "sector", "pe_ratio"]].head()
```

## Part 6: Combine Filter + Sort + Select

Chain operations together to answer financial questions.

```python
df.head(10)
```

```python
# TODO: Question 1 — What are the top 3 highest-yielding REITs?
#   reits = df[df["sector"].str.contains("REIT")]
#   reits.sort_values("dividend_yield", ascending=False)[["name", "sector", "dividend_yield", "price"]].head(3)
reits = df[df["sector"].str.contains("REIT")]
reits.sort_values("dividend_yield", ascending=False)[["name", "sector", "dividend_yield", "price"]].head(3)
```

```python
# TODO: Question 2 — Which banking stock has the lowest PE ratio?
#   banks = df[df["sector"] == "Banking"]
#   banks.sort_values("pe_ratio")[["name", "pe_ratio", "price"]].head(1)
banks = df[df["sector"] == "Banking"]
banks.sort_values("pe_ratio")[["name", "pe_ratio", "price"]].head(1)
```
