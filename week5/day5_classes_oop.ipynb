## Part 1: Your First Class

New concepts:
- `class ClassName:` — define a blueprint
- `def __init__(self, ...):` — the **constructor**, runs when you create an object
- `self.name = name` — store data on the object (called **attributes**)
- `self` — refers to "this particular object" (like "me" in English)

```python
# TODO: Create a Stock class
#
#   class Stock:
#       def __init__(self, ticker, name, price):
#           self.ticker = ticker
#           self.name = name
#           self.price = price
#
#   # Create two stocks
#   dbs = Stock("D05", "DBS Group", 41.20)
#   ocbc = Stock("O39", "OCBC Bank", 15.80)
#
#   # Access their attributes
#   print(f"{dbs.name}: ${dbs.price}")
#   print(f"{ocbc.name}: ${ocbc.price}")
#
# How it works:
#   Stock("D05", "DBS Group", 41.20) calls __init__ automatically.
#   self = the new object being created
#   self.ticker = "D05" stores "D05" on that object
#   Later, dbs.ticker retrieves it
```

## Part 2: Adding Methods

New concepts:
- A **method** is a function inside a class — it can access `self` (the object's data)
- `def describe(self):` — a method that uses `self.name`, `self.price`, etc.
- Call it with `dbs.describe()` — Python passes `dbs` as `self` automatically

```python
# TODO: Add methods to the Stock class
#
#   class Stock:
#       def __init__(self, ticker, name, price, dividend_yield):
#           self.ticker = ticker
#           self.name = name
#           self.price = price
#           self.dividend_yield = dividend_yield
#
#       def describe(self):
#           print(f"{self.ticker} | {self.name} | ${self.price} | Yield: {self.dividend_yield}%")
#
#       def annual_dividend(self):
#           return self.price * self.dividend_yield / 100
#
#   # Create stocks and call methods
#   dbs = Stock("D05", "DBS Group", 41.20, 5.1)
#   ocbc = Stock("O39", "OCBC Bank", 15.80, 5.4)
#
#   dbs.describe()
#   ocbc.describe()
#   print(f"\nDBS annual dividend per share: ${dbs.annual_dividend():.2f}")
#   print(f"OCBC annual dividend per share: ${ocbc.annual_dividend():.2f}")
#
# How it works:
#   dbs.describe()         — Python calls describe(self=dbs)
#   dbs.annual_dividend()  — Python calls annual_dividend(self=dbs)
#   Inside the method, self.price is dbs.price (41.20)
```

```python
class Stock:
    def __init__(self, ticker, name, price, dividend_yield):
        self.ticker = ticker
        self.name = name
        self.price = price
        self.dividend_yield = dividend_yield

    def describe(self):
        print(f"{self.ticker} | {self.name} | ${self.price} | Yield: {self.dividend_yield}%")

    def annual_dividend(self):
        return self.price * self.dividend_yield / 100
```

```python
#   # Create stocks and call methods
dbs = Stock("D05", "DBS Group", 41.20, 5.1)
ocbc = Stock("O39", "OCBC Bank", 15.80, 5.4)
```

```python
dbs.describe()
ocbc.describe()

print(f"\nDBS annual dividend per share: ${dbs.annual_dividend():.2f}")
print(f"OCBC annual dividend per share: ${ocbc.annual_dividend():.2f}")
```

## Part 3: Methods That Take Arguments

New concept:
- Methods can take extra arguments beyond `self`
- `def investment_income(self, shares):` — `self` is the stock, `shares` is how many you own

```python
# TODO: Add methods that take arguments
#
#   class Stock:
#       def __init__(self, ticker, name, price, dividend_yield, pe_ratio):
#           self.ticker = ticker
#           self.name = name
#           self.price = price
#           self.dividend_yield = dividend_yield
#           self.pe_ratio = pe_ratio
#
#       def describe(self):
#           print(f"{self.ticker} | {self.name} | ${self.price} | Yield: {self.dividend_yield}% | PE: {self.pe_ratio}")
#
#       def annual_dividend(self):
#           return self.price * self.dividend_yield / 100
#
#       def investment_income(self, shares):
#           return self.annual_dividend() * shares
#
#       def is_undervalued(self, max_pe=15):
#           return self.pe_ratio < max_pe
#
#   # Test it
#   dbs = Stock("D05", "DBS Group", 41.20, 5.1, 10.2)
#   dbs.describe()
#
#   income = dbs.investment_income(1000)
#   print(f"\n1000 shares of DBS → ${income:.2f}/year in dividends")
#
#   print(f"DBS undervalued (PE < 15)? {dbs.is_undervalued()}")
#   print(f"DBS undervalued (PE < 10)? {dbs.is_undervalued(10)}")
#
# New pattern:
#   investment_income calls self.annual_dividend() — methods can call other methods!
#   is_undervalued(max_pe=15) has a default argument — if you don't pass one, it uses 15
```

```python
class Stock:
    def __init__(self, ticker, name, price, dividend_yield, pe_ratio):
        self.ticker = ticker
        self.name = name
        self.price = price
        self.dividend_yield = dividend_yield
        self.pe_ratio = pe_ratio

    def describe(self):
        print(f"{self.ticker} | {self.name} | ${self.price} | Yield: {self.dividend_yield}% | PE: {self.pe_ratio}")

    def annual_dividend(self):
        return self.price * self.dividend_yield / 100

    def investment_income(self, shares):
        return self.annual_dividend() * shares

    def is_undervalued(self, max_pe=15):
        return self.pe_ratio < max_pe
```

```python
# Test it
dbs = Stock("D05", "DBS Group", 41.20, 5.1, 10.2)
dbs.describe()
#
income = dbs.investment_income(1000)
print(f"\n1000 shares of DBS → ${income:.2f}/year in dividends")
#
print(f"DBS undervalued (PE < 15)? {dbs.is_undervalued()}")
print(f"DBS undervalued (PE < 10)? {dbs.is_undervalued(10)}")
```

## Part 4: A Class That Holds a List

New concepts:
- A class can store a **list** of items as an attribute
- Methods can loop through that list, filter it, sort it
- This is how real libraries work — `ChromaDB` stores a collection of documents, `pandas` stores a table of rows

```python
# TODO: Build a Portfolio class that holds multiple stocks
#
#   class Portfolio:
#       def __init__(self, name):
#           self.name = name
#           self.stocks = []          # start with an empty list
#
#       def add_stock(self, ticker, name, price, dividend_yield):
#           stock = {"ticker": ticker, "name": name, "price": price, "dividend_yield": dividend_yield}
#           self.stocks.append(stock)
#           print(f"  Added {name}")
#
#       def show_all(self):
#           print(f"\n{self.name} ({len(self.stocks)} stocks):")
#           print("-" * 50)
#           for s in self.stocks:
#               print(f"  {s['ticker']} | {s['name']} | ${s['price']} | Yield: {s['dividend_yield']}%")
#
#       def total_value(self, shares_each=100):
#           total = sum(s["price"] * shares_each for s in self.stocks)
#           return total
#
#       def highest_yield(self):
#           best = max(self.stocks, key=lambda s: s["dividend_yield"])
#           return best
#
#   # Build a portfolio
#   sgx = Portfolio("SG Blue Chips")
#   sgx.add_stock("D05", "DBS Group", 41.20, 5.1)
#   sgx.add_stock("O39", "OCBC Bank", 15.80, 5.4)
#   sgx.add_stock("U11", "UOB", 34.50, 4.8)
#   sgx.add_stock("A17U", "CapitaLand Ascendas REIT", 2.78, 5.6)
#
#   sgx.show_all()
#   print(f"\nTotal value (100 shares each): ${sgx.total_value():,.2f}")
#   best = sgx.highest_yield()
#   print(f"Highest yield: {best['name']} ({best['dividend_yield']}%)")
```

```python
class Portfolio:
    def __init__(self, name):
        self.name = name
        self.stocks = []          # start with an empty list

    def add_stock(self, ticker, name, price, dividend_yield):
        stock = {"ticker": ticker, "name": name, "price": price, "dividend_yield": dividend_yield}
        self.stocks.append(stock)
        print(f"  Added {name}")

    def show_all(self):
        print(f"\n{self.name} ({len(self.stocks)} stocks):")
        print("-" * 50)
        for s in self.stocks:
              print(f"  {s['ticker']} | {s['name']} | ${s['price']} | Yield: {s['dividend_yield']}%")

    def total_value(self, shares_each=100):
        total = sum(s["price"] * shares_each for s in self.stocks)
        return total

    def highest_yield(self):
        best = max(self.stocks, key=lambda s: s["dividend_yield"])
        return best

  # Build a portfolio
sgx = Portfolio("SG Blue Chips")
sgx.add_stock("D05", "DBS Group", 41.20, 5.1)
sgx.add_stock("O39", "OCBC Bank", 15.80, 5.4)
sgx.add_stock("U11", "UOB", 34.50, 4.8)
sgx.add_stock("A17U", "CapitaLand Ascendas REIT", 2.78, 5.6)

sgx.show_all()
print(f"\nTotal value (100 shares each): ${sgx.total_value():,.2f}")
best = sgx.highest_yield()
print(f"Highest yield: {best['name']} ({best['dividend_yield']}%)")
```

## Part 5: StockAnalyzer Class — Load CSV + Analyze

New concept:
- A class can use `pandas` to load data in `__init__`
- Methods become **reusable analysis tools** — call them anytime on the loaded data
- This is the pattern you'll use for every data project going forward

```python
# TODO: Build a StockAnalyzer class that loads CSV data
#
import pandas as pd

class StockAnalyzer:
  def __init__(self, csv_path):
      self.df = pd.read_csv(csv_path)
      print(f"Loaded {len(self.df)} stocks from {csv_path}")

  def top_by_yield(self, n=5):
      top = self.df.sort_values("dividend_yield", ascending=False).head(n)
      return top[["name", "sector", "dividend_yield", "price"]]

  def sector_summary(self):
      return self.df.groupby("sector").agg(
          count=("name", "count"),
          avg_yield=("dividend_yield", "mean"),
          avg_pe=("pe_ratio", "mean")
      ).sort_values("avg_yield", ascending=False)

  def find_undervalued(self, max_pe=12, min_yield=4.0):
      matches = self.df[
          (self.df["pe_ratio"] < max_pe) &
          (self.df["dividend_yield"] > min_yield)
      ]
      return matches[["name", "sector", "pe_ratio", "dividend_yield", "price"]]

# Use it
analyzer = StockAnalyzer("data/sgx_cleaned.csv")

print("\nTop 5 by dividend yield:")
print(analyzer.top_by_yield())

print("\nSector summary:")
print(analyzer.sector_summary())

print("\nUndervalued stocks (PE < 12, yield > 4%):")
print(analyzer.find_undervalued())
```

## Part 6: Add AI Analysis to the Class

New concept:
- A class can combine **data** (pandas) and **AI** (Claude) in one object
- `__init__` sets up both the data and the API client
- One method prepares the data, another sends it to Claude
- This is the pattern for the Saturday dashboard project

```python
# TODO: Add Claude AI commentary to the StockAnalyzer
#
import anthropic

class SmartAnalyzer:
  def __init__(self, csv_path):
      self.df = pd.read_csv(csv_path)
      self.client = anthropic.Anthropic()
      self.model = "claude-sonnet-5"
      print(f"Loaded {len(self.df)} stocks")

  def get_summary_text(self):
      top = self.df.sort_values("dividend_yield", ascending=False).head(5)
      sector = self.df.groupby("sector")["dividend_yield"].mean()
      summary = f"Top yielding stocks:\n{top[['name','sector','dividend_yield']].to_string()}\n\n"
      summary += f"Average yield by sector:\n{sector.to_string()}"
      return summary

  def ai_commentary(self):
      summary = self.get_summary_text()
      message = self.client.messages.create(
          model=self.model,
          max_tokens=300,
          system="You are a Singapore equity analyst. Give a brief 3-4 sentence commentary.",
          messages=[{"role": "user", "content": f"Comment on this SGX market data:\n\n{summary}"}]
      )
      return message.content[0].text

# Use it
smart = SmartAnalyzer("data/sgx_cleaned.csv")

print("=== Data Summary ===")
print(smart.get_summary_text())

print("\n=== AI Commentary ===")
print(smart.ai_commentary())
```
