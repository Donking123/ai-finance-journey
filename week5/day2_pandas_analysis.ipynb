## Part 2: groupby — Split Data by Category

New concepts:
- `df.groupby("sector")` — group rows that share the same sector
- Think of it like: "for each sector, do something"
- By itself, groupby does nothing visible — you need to add an aggregation (.mean(), .count(), etc.)

How it works:
1. **Split** — pandas divides rows into groups (Banking, Technology, etc.)
2. **Apply** — run a function on each group (mean, sum, count)
3. **Combine** — merge results into a new DataFrame

## Part 4: agg() — Multiple Aggregations at Once

New concepts:
- `.agg()` lets you apply MULTIPLE functions to MULTIPLE columns in one call
- Pass a dictionary: `{"column": "function"}` or `{"column": ["func1", "func2"]}`
- Much more powerful than calling .mean(), .sum() one at a time

```python
df
```

```python
sector_summary = df.groupby("sector").agg({
    "name": "count",
    "dividend_yield" : "mean",
    "market_cap_m": "sum",
    "pe_ratio" : "mean"
})
sector_summary
```

```python
# TODO: Add a "yield_category" column based on dividend yield
#   High yield = above 5%, Medium = 3-5%, Low = below 3%
#
#   def classify_yield(y):
#       if y >= 5.0:
#           return "High"
#       elif y >= 3.0:
#           return "Medium"
#       else:
#           return "Low"
#
#   df["yield_category"] = df["dividend_yield"].apply(classify_yield)
#   df[["name", "dividend_yield", "yield_category"]]
```

```python
def classify_yield(y):
    if y >= 5.0:
        return "High"
    elif y >= 3.0:
        return "Medium"
    else:
        return "Low"
```

```python
df["yield_category"] = df["dividend_yield"].apply(classify_yield)
```

```python
df[["name", "dividend_yield", "yield_category"]]
```
